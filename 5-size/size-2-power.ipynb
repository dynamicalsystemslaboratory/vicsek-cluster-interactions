```julia
include( "../model.jl" )
include( "../cluster.jl" )
include( "../prob.jl" )

include( "size-0-params.jl" )

using Optim
using ThreadsX

home, base = expfolder( "system" )
```

```julia
# Hyper parameters.
N = 128

# Create folder and extract critical noise.
folderdata = [findfolder( N, ρ, s; base=base, crit=true ) for ρ ∈ ρlist, s ∈ slist]

# Import cluster frequency data.
Fdata = [F for F ∈ vec.( readdlm.( folderdata.*"cluster-frequency.txt" ) )]

# Initialize data variables.
λdata = [10^optimize( createzipfloss( N, Fdata[i,j] ),  [log10( 1.5 )] ).minimizer[1]
    for i ∈ 1:Nρ, j ∈ 1:Ns];
```

```julia
# Save parameter data.
if true
    # Save the cluster change parameters for each parameter case.
    for i ∈ 1:Nρ, j ∈ 1:Ns
        writedlm( folderdata[i,j]*"power-law-hat.txt", λdata[i,j] )
    end
end
```
